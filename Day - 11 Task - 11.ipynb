```python
# ============================================================
# ABTalks (Season-3) - AI Engineering
# Task Day-11 : Text Similarity using Cosine Similarity
# Tool/Tech : Scikit-learn (sklearn)
# Problem Statement:
# Compute cosine similarity between two documents
# ============================================================

# -----------------------------
# Import Required Libraries
# -----------------------------
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

# ============================================================
# STEP 1 : Define Two Documents
# ============================================================

document_1 = """
Artificial Intelligence is transforming modern industries by enabling
machines to learn from data and make intelligent decisions.
"""

document_2 = """
AI is changing industries through machine learning, data analysis,
and smart decision-making systems.
"""

# Store documents in a list
documents = [document_1, document_2]

# ============================================================
# STEP 2 : Text Vectorization using TF-IDF
# ============================================================

# Create TF-IDF Vectorizer Object
vectorizer = TfidfVectorizer(
    stop_words='english',
    lowercase=True
)

# Convert text documents into TF-IDF matrix
tfidf_matrix = vectorizer.fit_transform(documents)

# ============================================================
# STEP 3 : Compute Cosine Similarity
# ============================================================

similarity_matrix = cosine_similarity(tfidf_matrix)

# Extract similarity score between Document 1 and Document 2
similarity_score = similarity_matrix[0][1]

# ============================================================
# STEP 4 : Display Results
# ============================================================

print("\n" + "="*60)
print("TEXT SIMILARITY ANALYSIS USING COSINE SIMILARITY")
print("="*60)

print("\nDocument 1:\n")
print(document_1)

print("\nDocument 2:\n")
print(document_2)

print("\n" + "="*60)
print(f"Cosine Similarity Score : {similarity_score:.4f}")
print("="*60)

# ============================================================
# STEP 5 : Similarity Interpretation
# ============================================================

if similarity_score > 0.75:
    print("\nResult : Documents are Highly Similar")
elif similarity_score > 0.40:
    print("\nResult : Documents are Moderately Similar")
else:
    print("\nResult : Documents are Less Similar")

# ============================================================
# STEP 6 : Display TF-IDF Features
# ============================================================

feature_names = vectorizer.get_feature_names_out()

# Convert sparse matrix to dataframe
tfidf_dataframe = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=feature_names,
    index=["Document_1", "Document_2"]
)

print("\n" + "="*60)
print("TF-IDF FEATURE MATRIX")
print("="*60)

print(tfidf_dataframe)

# ============================================================
# STEP 7 : Show Complete Similarity Matrix
# ============================================================

similarity_df = pd.DataFrame(
    similarity_matrix,
    columns=["Document_1", "Document_2"],
    index=["Document_1", "Document_2"]
)

print("\n" + "="*60)
print("COSINE SIMILARITY MATRIX")
print("="*60)

print(similarity_df)

# ============================================================
# FINAL INSIGHT
# ============================================================

print("\n" + "="*60)
print("FINAL INSIGHT")
print("="*60)

print("""
Cosine Similarity measures the semantic similarity between two documents.
The value ranges from 0 to 1:

0   -> Completely Different
1   -> Exactly Similar

TF-IDF helps in converting textual data into numerical vectors,
and cosine similarity computes the angle between these vectors
to determine how closely related the documents are.
""")

# ============================================================
# END OF PROJECT
# ============================================================
```
